```python
import math

def left_rectangles(f, a, b, n):
    """
    Вычисление интеграла методом левых прямоугольников.
    :param f: функция f(x)
    :param a: левая граница
    :param b: правая граница
    :param n: количество разбиений
    :return: значение интеграла
    """
    h = (b - a) / n
    integral = 0.0
    for i in range(n):
        x_i = a + i * h
        integral += f(x_i)
    return integral * h

def f(x):
    return math.sin(x)

a = 0
b = math.pi

n = 10
epsilon = 1e-10

print(f"Интеграл ∫_{a}^{b} sin(x) dx")
print(f"Эпсилон {epsilon}")
print(f"Точное значение: {math.cos(a) - math.cos(b):.8f}")
print("-" * 50)

print(f"{'n':<8} {'I_n':<20} {'|I_n - I_2n|':<15} {'Достигнута'}")
print("-" * 60)

while True:
    I_n = left_rectangles(f, a, b, n)
    I_2n = left_rectangles(f, a, b, n * 2)
    diff = abs(I_n - I_2n)

    print(f"{n:<8} {I_n:<20} {diff:<15.8e} {'Да' if diff < epsilon else 'Нет'}")

    if diff < epsilon:
        print(f"\nПриближённое значение интеграла: {I_2n:.8f}")
        print(f"Количество разбиений: {n * 2}")
        break

    n *= 2

    if n > 10**9:
        print("ошибка")
        break
```
