```python
import numpy as np
import time

def f(x, k):
    return (k - a) * x + np.exp((k - b) * (k - c) * x)

# Аналитическое решение
def analytical_min(k):
    beta = (k - b) * (k - c)
    alpha = k - a

    if beta == 0:
        return None

    condition = -alpha / beta
    if condition <= 0:
        return None

    return np.log(condition) / beta

# Методы оптимизации
def dihotomy(f, a, b, k, tol=1e-6):
    start = time.time()

    i = 0
    delta = tol / 10
    while abs(a - b) > tol:
        mid = (a + b) / 2

        x1 = mid - delta
        x2 = mid + delta

        f1 = f(x1, k)
        f2 = f(x2, k)

        if f1 < f2:
            b = mid
        else:
            a = mid
        i += 1

    return (a + b) / 2, i, time.time() - start

def golden_section(f, a, b, k, tol=1e-6):
    start = time.time()
    phi = (3 - 5**0.5) / 2
    x1 = a + phi * (b - a)
    x2 = b - phi * (b - a)

    f1, f2 = f(x1, k), f(x2, k)

    for i in range(100):
        if abs(b - a) < tol:
            return (a + b) / 2, i+1, time.time() - start

        if f1 < f2:
            b = x2
            x2 = x1
            f2 = f1
            x1 = a + phi * (b - a)
            f1 = f(x1, k)
        else:
            a = x1
            x1 = x2
            f1 = f2
            x2 = b - phi * (b - a)
            f2 = f(x2, k)

    return (a + b) / 2, 100, time.time() - start

def fibonacci(f, a, b, k, tol=1e-6):
    start = time.time()

    # Генерация чисел Фибоначчи
    fib = [1, 1]
    while fib[-1] < (b - a) / tol:
        fib.append(fib[-1] + fib[-2])
    n = len(fib) - 1

    # Начальные точки
    x1 = a + fib[n-2] / fib[n] * (b - a)
    x2 = a + fib[n-1] / fib[n] * (b - a)
    f1, f2 = f(x1, k), f(x2, k)

    for i in range(1, n):
        if abs(b - a) < tol:
            return (a + b) / 2, i, time.time() - start

        if f1 < f2:
            b = x2
            x2 = x1
            f2 = f1
            x1 = a + fib[n-i-1] / fib[n-i+1] * (b - a)
            f1 = f(x1, k)
        else:
            a = x1
            x1 = x2
            f1 = f2
            x2 = a + fib[n-i] / fib[n-i+1] * (b - a)
            f2 = f(x2, k)

    return (a + b) / 2, n, time.time() - start

a, b, c = 7, 3, 5
k = 2
beta = (k - b) * (k - c)
alpha = k - a
X_LEFT, X_RIGHT = -5, 5

if beta == 0:
    print(f"beta = 0, функция линейна, минимума нет")
elif -alpha/beta <= 0:
    print(f"-alpha/beta = {-alpha/beta:.3f} ≤ 0, минимума нет")
else:
    x_exact = analytical_min(k)
    print(f"k = {k} \n(k - {b}) * (k - {c}) = {alpha:.2f} > 0")
    print(f"Аналитически: x* = {x_exact:.6f}, f(x*) = {f(x_exact, k):.6f}")

    print(f"\n{'Метод':<20} {'x_min':<12} {'f(x)':<12} {'Итер.':<8} {'Погрешность':<10}")
    print("-" * 65)

    methods = [
        ("Дихотомия", lambda: dihotomy(f, X_LEFT, X_RIGHT, k)),
        ("Золотого сечения", lambda: golden_section(f, X_LEFT, X_RIGHT, k)),
        ("Фибоначчи", lambda: fibonacci(f, X_LEFT, X_RIGHT, k))
    ]

    results = []
    for name, method in methods:
        x, iters, t = method()
        fx = f(x, k)
        error = abs(x - x_exact)
        results.append((name, x, fx, iters, t, error))
        print(f"{name:<20} {x:<12.6f} {fx:<12.6f} {iters:<8} {error:<10.2e}")

    best = min(results, key=lambda x: x[5])
    print(f"\nЛучший по точности: {best[0]} (погрешность {best[5]:.2e})")
```
